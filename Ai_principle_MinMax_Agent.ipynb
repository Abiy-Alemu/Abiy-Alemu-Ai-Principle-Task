3. Problem Formulation as a Game

          A game in AI consists of:

          Component	Description
          Players	MAX (AI agent) and MIN (Opponent)
          Initial State	Empty game board
          Actions	Legal moves allowed in the game
          Result	New board after a move
          Terminal Test	Win, lose, or draw
          Utility Function	+1 (win), −1 (lose), 0 (draw)

          We use Tic-Tac-Toe as the example game because it clearly demonstrates Minimax.

4. Game Tree Representation

        Each node represents a board state

        Edges represent legal moves

        Leaf nodes are terminal states

        Values are propagated upward using Minimax

            Root (MAX)
           /          \
        MIN            MIN
       /   \          /   \
     MAX   MAX      MAX   MAX


5. The Minimax Algorithm

Minimax assumes:

    MAX tries to maximize the score

    MIN tries to minimize the score

Pseudocode


function MINIMAX(state, isMax):
    if state is terminal:
        return utility(state)

    if isMax:
        best = -∞
        for each move in state:
            best = max(best, MINIMAX(result(move), false))
        return best
    else:
        best = +∞
        for each move in state:
            best = min(best, MINIMAX(result(move), true))
        return best
