    Search Class Implementation

```python
from collections import deque

class EthiopiaSearch:
    def __init__(self, graph, start, goal, strategy):
        self.graph = graph
        self.start = start
        self.goal = goal
        self.strategy = strategy.lower()

    def search(self):
        if self.strategy == "bfs":
            return self.bfs()
        elif self.strategy == "dfs":
            return self.dfs()
        else:
            raise ValueError("Strategy must be 'bfs' or 'dfs'")

    def bfs(self):
        queue = deque([[self.start]])
        visited = set()

        while queue:
            path = queue.popleft()
            node = path[-1]

            if node == self.goal:
                return path

            if node not in visited:
                visited.add(node)
                for neighbor in self.graph.get(node, []):
                    new_path = list(path)
                    new_path.append(neighbor)
                    queue.append(new_path)

        return None

    def dfs(self):
        stack = [[self.start]]
        visited = set()

        while stack:
            path = stack.pop()
            node = path[-1]

            if node == self.goal:
                return path

            if node not in visited:
                visited.add(node)
                for neighbor in self.graph.get(node, []):
                    new_path = list(path)
                    new_path.append(neighbor)
                    stack.append(new_path)

        return None



search_bfs = EthiopiaSearch(
    ethiopia_graph,
    start="Addis Ababa",
    goal="Harar",
    strategy="bfs"
)

search_dfs = EthiopiaSearch(
    ethiopia_graph,
    start="Addis Ababa",
    goal="Harar",
    strategy="dfs"
)

print("BFS Path:", search_bfs.search())
print("DFS Path:", search_dfs.search())

```

2.2 Uniform Cost Search from Addis Ababa to Lalibela

```python
import heapq

def uniform_cost_search(graph, start, goal):
    """
    Uniform Cost Search algorithm to find the path with minimum cost
    """
    # Priority queue: (cost, path)
    frontier = [(0, [start])]
    visited = set()
    explored_nodes = []

    while frontier:
        # Get the node with minimum cost
        cost, path = heapq.heappop(frontier)
        current_node = path[-1]

        if current_node in visited:
            continue

        visited.add(current_node)
        explored_nodes.append(current_node)

        # Check if we reached the goal
        if current_node == goal:
            return {
                'path': path,
                'cost': cost,
                'explored_nodes': explored_nodes
            }

        # Explore neighbors
        for neighbor, step_cost in graph.get(current_node, []):
            if neighbor not in visited:
                new_cost = cost + step_cost
                new_path = path + [neighbor]
                heapq.heappush(frontier, (new_cost, new_path))

    return None  # No path found

# Run UCS from Addis Ababa to Lalibela
result = uniform_cost_search(ethiopia_graph, 'Addis Ababa', 'Lalibela')

if result:
    print("=" * 60)
    print("UNIFORM COST SEARCH: Addis Ababa to Lalibela")
    print("=" * 60)
    print(f"Path found: {' -> '.join(result['path'])}")
    print(f"Total cost: {result['cost']}")
    print(f"\nNodes explored in order: {', '.join(result['explored_nodes'][:10])}...")
    print(f"Total nodes explored: {len(result['explored_nodes'])}")
else:
    print("No path found from Addis Ababa to Lalibela")
```

    2.3 Customized UCS to Visit All Goal States

```python
import heapq
from itertools import permutations

def customized_ucs_multi_goal(graph, start, goals):
    """
    Customized Uniform Cost Search to visit all goal states preserving local optimum
    This finds a path that visits all goals in an optimal order
    """

    def shortest_path_between(graph, start, end):
        """Find shortest path between two nodes using UCS"""
        frontier = [(0, [start])]
        visited = set()

        while frontier:
            cost, path = heapq.heappop(frontier)
            current = path[-1]

            if current in visited:
                continue

            visited.add(current)

            if current == end:
                return cost, path

            for neighbor, step_cost in graph.get(current, []):
                if neighbor not in visited:
                    new_cost = cost + step_cost
                    new_path = path + [neighbor]
                    heapq.heappush(frontier, (new_cost, new_path))

        return float('inf'), None

    # Add start to goals list temporarily for path computation
    all_points = [start] + goals

    # Precompute shortest paths between all points
    print("Precomputing distances between all points...")
    distances = {}
    paths = {}

    total_pairs = len(all_points) * (len(all_points) - 1)
    computed = 0

    for i, city1 in enumerate(all_points):
        for city2 in all_points:
            if city1 != city2:
                key = (city1, city2)
                cost, path = shortest_path_between(graph, city1, city2)
                distances[key] = cost
                paths[key] = path

                computed += 1
                if computed % 10 == 0:
                    print(f"Progress: {computed}/{total_pairs} pairs computed")

    print("\n" + "=" * 70)
    print("CUSTOMIZED UCS: Visit All Goal States")
    print("=" * 70)
    print(f"Start: {start}")
    print(f"Goals to visit: {', '.join(goals)}")
    print("-" * 70)

    # Try all permutations of goals to find optimal order
    best_cost = float('inf')
    best_order = None
    best_full_path = None

    print("Finding optimal visiting order...")
    for i, perm in enumerate(permutations(goals)):
        if i % 100 == 0:  # Print progress every 100 permutations
            print(f"Checking permutation {i}...")

        # Calculate total cost for this permutation
        total_cost = 0
        full_path = [start]
        current = start

        valid = True
        for next_goal in perm:
            key = (current, next_goal)
            if key not in distances or distances[key] == float('inf'):
                valid = False
                break

            total_cost += distances[key]
            # Add path (excluding the starting point to avoid duplication)
            path_segment = paths[key][1:]
            full_path.extend(path_segment)
            current = next_goal

        if valid and total_cost < best_cost:
            best_cost = total_cost
            best_order = perm
            best_full_path = full_path

    print("\n" + "=" * 70)
    print("OPTIMAL SOLUTION FOUND")
    print("=" * 70)
    print(f"Optimal visiting order: {start} -> {' -> '.join(best_order)}")
    print(f"Total travel cost: {best_cost}")
    print(f"\nComplete path:")
    path_str = ' -> '.join(best_full_path)
    # Wrap long path for readability
    if len(path_str) > 80:
        parts = path_str.split(' -> ')
        wrapped_lines = []
        current_line = []
        current_length = 0

        for part in parts:
            if current_length + len(part) + 3 > 80:
                wrapped_lines.append(' -> '.join(current_line))
                current_line = [part]
                current_length = len(part)
            else:
                current_line.append(part)
                current_length += len(part) + 3

        if current_line:
            wrapped_lines.append(' -> '.join(current_line))

        for line in wrapped_lines:
            print(f"  {line}")
    else:
        print(f"  {path_str}")

    print(f"\nNumber of cities visited: {len(best_full_path)}")

    return {
        'order': best_order,
        'cost': best_cost,
        'full_path': best_full_path,
        'distances': distances,
        'paths': paths
    }

# Define goals based on the problem statement
goals = ['Axum', 'Gondar', 'Lalibela', 'Babile', 'Jimma', 'Bale', 'Sof Oumer', 'Arba Minch']

# Run the customized algorithm
result = customized_ucs_multi_goal(ethiopia_graph, 'Addis Ababa', goals)

# Optional: Show distance matrix for key cities
print("\n" + "=" * 70)
print("DISTANCE MATRIX (Selected Cities)")
print("=" * 70)

key_cities = ['Addis Ababa', 'Axum', 'Gondar', 'Lalibela', 'Jimma', 'Arba Minch']
print(f"{'':15}", end='')
for city in key_cities:
    print(f"{city[:10]:>12}", end='')
print()

for city1 in key_cities:
    print(f"{city1[:15]:15}", end='')
    for city2 in key_cities:
        if city1 == city2:
            print(f"{'0':>12}", end='')
        else:
            key = (city1, city2)
            if key in result['distances']:
                print(f"{result['distances'][key]:>12}", end='')
            else:
                print(f"{'inf':>12}", end='')
    print()
```

```python
import heapq
import math
from typing import List, Tuple, Dict, Optional, Set

class TravelingEthiopiaAStar:
    """
    A* Search implementation for finding optimal path from Addis Ababa to Moyale
    """

    def __init__(self):
        # Graph structure with backward costs (actual travel costs)
        self.graph = {
            # Central Ethiopia
            'Addis Ababa': [('Ambo', 1), ('Buta Jirra', 2), ('Adama', 3), ('Debre Berhan', 4)],
            'Ambo': [('Addis Ababa', 1), ('Wolkite', 2), ('Nekemte', 3)],
            'Buta Jirra': [('Addis Ababa', 2), ('Wolkite', 1), ('Worabe', 3)],
            'Adama': [('Addis Ababa', 3), ('Matahara', 2), ('Assella', 3)],
            'Debre Berhan': [('Addis Ababa', 4), ('Debre Sina', 2)],

            # Western routes
            'Wolkite': [('Ambo', 2), ('Buta Jirra', 1), ('Jimma', 4)],
            'Nekemte': [('Ambo', 3), ('Gimbi', 4)],
            'Worabe': [('Buta Jirra', 3), ('Hossana', 2)],
            'Jimma': [('Wolkite', 4), ('Bonga', 3), ('Bedelle', 4)],
            'Gimbi': [('Nekemte', 4), ('Dembidolo', 5)],
            'Bonga': [('Jimma', 3), ('Tepi', 4), ('Mizan Teferi', 4)],
            'Bedelle': [('Jimma', 4), ('Gore', 3)],
            'Tepi': [('Bonga', 4), ('Gambella', 5)],
            'Mizan Teferi': [('Bonga', 4)],
            'Gore': [('Bedelle', 3), ('Gambella', 4)],
            'Gambella': [('Tepi', 5), ('Gore', 4)],

            # Southern routes
            'Hossana': [('Worabe', 2), ('Wolaita Sodo', 3)],
            'Wolaita Sodo': [('Hossana', 3), ('Arba Minch', 4)],
            'Arba Minch': [('Wolaita Sodo', 4), ('Konso', 3)],
            'Konso': [('Arba Minch', 3), ('Yabello', 4)],
            'Yabello': [('Konso', 4), ('Moyale', 5)],

            # Eastern routes
            'Matahara': [('Adama', 2), ('Awash', 2)],
            'Assella': [('Adama', 3), ('Assasa', 3)],
            'Awash': [('Matahara', 2), ('Chiro', 3)],
            'Assasa': [('Assella', 3), ('Dodolla', 2)],
            'Chiro': [('Awash', 3), ('Dire Dawa', 4)],
            'Dodolla': [('Assasa', 2), ('Bale', 3)],
            'Dire Dawa': [('Chiro', 4), ('Harar', 2)],
            'Harar': [('Dire Dawa', 2), ('Babile', 3)],
            'Babile': [('Harar', 3), ('Jijiga', 3)],
            'Jijiga': [('Babile', 3), ('Degehabur', 4)],
            'Degehabur': [('Jijiga', 4), ('Kebri Dehar', 5)],
            'Kebri Dehar': [('Degehabur', 5), ('Gode', 6)],
            'Gode': [('Kebri Dehar', 6), ('Dollo', 7)],
            'Dollo': [('Gode', 7)],

            # Northern routes
            'Debre Sina': [('Debre Berhan', 2), ('Kemise', 3)],
            'Kemise': [('Debre Sina', 3), ('Dessie', 3)],
            'Dessie': [('Kemise', 3), ('Woldia', 3)],
            'Woldia': [('Dessie', 3), ('Lalibela', 2), ('Samara', 4)],
            'Lalibela': [('Woldia', 2), ('Sekota', 3)],
            'Sekota': [('Lalibela', 3), ('Mekelle', 4)],
            'Mekelle': [('Sekota', 4), ('Adigrat', 4), ('Adwa', 3)],
            'Adigrat': [('Mekelle', 4), ('Axum', 3)],
            'Adwa': [('Mekelle', 3), ('Axum', 2)],
            'Axum': [('Adwa', 2), ('Adigrat', 3), ('Shire', 3)],
            'Shire': [('Axum', 3), ('Humera', 4)],
            'Humera': [('Shire', 4), ('Gondar', 5)],
            'Gondar': [('Humera', 5), ('Debark', 3)],
            'Debark': [('Gondar', 3)],

            # Central highlands
            'Bale': [('Dodolla', 3), ('Goba', 3), ('Sof Oumer', 4)],
            'Goba': [('Bale', 3)],
            'Sof Oumer': [('Bale', 4), ('Goba', 5)],

            # Far north
            'Samara': [('Woldia', 4), ('Fanti Rasu', 5)],
            'Fanti Rasu': [('Samara', 5), ('Killbet', 6)],
            'Killbet': [('Fanti Rasu', 6)],

            # Goal state
            'Moyale': [('Yabello', 5)]
        }

        # Heuristic values (estimated straight-line/crow-flies distance to Moyale)
        # These are reasonable estimates based on geographical intuition
        self.heuristics = {
            # Goal state
            'Moyale': 0,

            # Southern region (closest to Moyale)
            'Yabello': 5,
            'Konso': 9,
            'Arba Minch': 13,
            'Wolaita Sodo': 17,
            'Hossana': 20,

            # Central region
            'Addis Ababa': 40,
            'Ambo': 42,
            'Buta Jirra': 38,
            'Adama': 35,
            'Debre Berhan': 45,
            'Wolkite': 35,
            'Worabe': 30,
            'Jimma': 45,
            'Bonga': 48,
            'Bedelle': 52,

            # Eastern region
            'Matahara': 40,
            'Assella': 32,
            'Awash': 45,
            'Assasa': 28,
            'Chiro': 48,
            'Dodolla': 25,
            'Dire Dawa': 55,
            'Harar': 58,
            'Babile': 60,
            'Bale': 22,
            'Goba': 24,
            'Sof Oumer': 26,

            # Northern region (far from Moyale)
            'Dessie': 55,
            'Woldia': 58,
            'Lalibela': 62,
            'Sekota': 65,
            'Mekelle': 70,
            'Axum': 75,
            'Gondar': 72,
            'Debre Sina': 48,
            'Kemise': 52,

            # Far eastern
            'Jijiga': 65,
            'Degehabur': 70,
            'Gode': 80,

            # Default for any unspecified nodes
            'DEFAULT': 50
        }

    def get_heuristic(self, node: str) -> int:
        """Get heuristic value for a node (estimated cost to Moyale)"""
        return self.heuristics.get(node, self.heuristics['DEFAULT'])

    def astar_search(self, start: str = 'Addis Ababa', goal: str = 'Moyale') -> Dict:
        """
        Perform A* search from start to goal

        Returns:
            Dictionary containing path, cost, and search statistics
        """

        # Priority queue: (f_score, g_score, path)
        # f = g + h (actual cost + heuristic)
        frontier = [(self.get_heuristic(start), 0, [start])]

        # Track visited nodes with their best g_scores
        visited = {}  # node -> best g_score
        explored_nodes = []  # Order of exploration

        # Statistics
        nodes_expanded = 0
        max_frontier_size = 0

        print(f"\n{'='*70}")
        print(f"A* SEARCH: {start} → {goal}")
        print(f"{'='*70}\n")

        while frontier:
            # Update max frontier size
            max_frontier_size = max(max_frontier_size, len(frontier))

            # Get node with minimum f_score
            f_current, g_current, path = heapq.heappop(frontier)
            current_node = path[-1]

            # Skip if we've found a better path to this node
            if current_node in visited and visited[current_node] <= g_current:
                continue

            # Mark as visited with this g_score
            visited[current_node] = g_current
            explored_nodes.append(current_node)
            nodes_expanded += 1

            # Print progress occasionally
            if nodes_expanded % 10 == 0 or nodes_expanded < 5:
                print(f"Expanding: {current_node:15} | g={g_current:3} | h={self.get_heuristic(current_node):3} | f={f_current:3}")

            # Check if we reached the goal
            if current_node == goal:
                print(f"\n{'✓'*50}")
                print(f"GOAL REACHED!")
                print(f"{'✓'*50}")

                return {
                    'path': path,
                    'cost': g_current,
                    'nodes_expanded': nodes_expanded,
                    'explored_nodes': explored_nodes,
                    'max_frontier_size': max_frontier_size,
                    'success': True
                }

            # Explore neighbors
            for neighbor, step_cost in self.graph.get(current_node, []):
                # Calculate new g score
                new_g = g_current + step_cost

                # If we haven't visited neighbor or found better path
                if neighbor not in visited or new_g < visited.get(neighbor, float('inf')):
                    # Calculate f score: f = g + h
                    h_neighbor = self.get_heuristic(neighbor)
                    new_f = new_g + h_neighbor

                    # Create new path
                    new_path = path + [neighbor]

                    # Add to frontier
                    heapq.heappush(frontier, (new_f, new_g, new_path))

        # No path found
        return {
            'path': None,
            'cost': float('inf'),
            'nodes_expanded': nodes_expanded,
            'explored_nodes': explored_nodes,
            'max_frontier_size': max_frontier_size,
            'success': False
        }

    def print_path_details(self, result: Dict) -> None:
        """Print detailed information about the found path"""

        if not result['success']:
            print("\n❌ No path found from Addis Ababa to Moyale")
            return

        print("\n" + "="*70)
        print("PATH DETAILS")
        print("="*70)

        path = result['path']
        total_cost = result['cost']

        print(f"\n📋 Complete path ({len(path)} cities):")
        print("-" * 50)

        # Print path with cumulative costs
        cumulative_cost = 0
        for i, city in enumerate(path):
            if i == 0:
                print(f"{i+1:2}. {city:15} (Start)")
            else:
                # Find the cost from previous city
                prev_city = path[i-1]
                for neighbor, cost in self.graph.get(prev_city, []):
                    if neighbor == city:
                        cumulative_cost += cost
                        print(f"{i+1:2}. {city:15} | +{cost:2} = {cumulative_cost:3} total")
                        break

        print("-" * 50)
        print(f"\n📊 Statistics:")
        print(f"   • Total path cost: {total_cost}")
        print(f"   • Nodes expanded: {result['nodes_expanded']}")
        print(f"   • Max frontier size: {result['max_frontier_size']}")

        # Verify heuristic admissibility
        print(f"\n🔍 Heuristic Analysis:")
        print(f"   • First node heuristic (Addis Ababa): {self.get_heuristic('Addis Ababa')}")
        print(f"   • Actual path cost: {total_cost}")
        if self.get_heuristic('Addis Ababa') <= total_cost:
            print(f"   ✓ Heuristic is admissible (h ≤ actual cost)")
        else:
            print(f"   ⚠ Heuristic overestimates! Not admissible")

    def visualize_search_progress(self, result: Dict) -> None:
        """Visualize the search progress"""

        if not result['success']:
            return

        print("\n" + "="*70)
        print("SEARCH PROGRESS VISUALIZATION")
        print("="*70)

        explored = result['explored_nodes']
        path = result['path']

        # Show exploration order
        print(f"\n🔍 Exploration order (first 20 nodes):")
        for i, node in enumerate(explored[:20]):
            if node in path:
                print(f"   {i+1:2}. {node:15} ✓ (on path)")
            else:
                print(f"   {i+1:2}. {node:15}")

        if len(explored) > 20:
            print(f"   ... and {len(explored) - 20} more nodes")

        # Show path in context
        print(f"\n🛣️  Path found (simplified):")
        print(f"   {' → '.join(path)}")

        # Calculate branching factor
        print(f"\n🌳 Search Tree Statistics:")
        print(f"   • Total nodes explored: {len(explored)}")
        print(f"   • Path length: {len(path)}")
        print(f"   • Effective branching factor: ~{len(explored) ** (1/len(path)):.2f}")


# Example usage and demonstration
def main():
    # Create search instance
    ethiopia_search = TravelingEthiopiaAStar()

    print("🌍 TRAVELING ETHIOPIA SEARCH PROBLEM")
    print("="*70)
    print("Finding optimal path from Addis Ababa to Moyale using A*")

    # Perform A* search
    result = ethiopia_search.astar_search('Addis Ababa', 'Moyale')

    # Print detailed results
    ethiopia_search.print_path_details(result)
    ethiopia_search.visualize_search_progress(result)

    # Compare with alternative path through different regions
    print("\n" + "="*70)
    print("ALTERNATIVE PATH ANALYSIS")
    print("="*70)

    # Southern route (likely optimal)
    southern_route = ['Addis Ababa', 'Buta Jirra', 'Worabe', 'Hossana',
                      'Wolaita Sodo', 'Arba Minch', 'Konso', 'Yabello', 'Moyale']
    southern_cost = 0
    for i in range(len(southern_route)-1):
        for neighbor, cost in ethiopia_search.graph.get(southern_route[i], []):
            if neighbor == southern_route[i+1]:
                southern_cost += cost
                break

    print(f"\n🛣️  Southern Route (through Rift Valley):")
    print(f"   {' → '.join(southern_route)}")
    print(f"   Total cost: {southern_cost}")

    # Central route through Bale
    central_route = ['Addis Ababa', 'Adama', 'Assella', 'Assasa',
                     'Dodolla', 'Bale', 'Sof Oumer', 'Goba', 'Yabello', 'Moyale']
    central_cost = 0
    for i in range(len(central_route)-1):
        for neighbor, cost in ethiopia_search.graph.get(central_route[i], []):
            if neighbor == central_route[i+1]:
                central_cost += cost
                break

    print(f"\n🛣️  Central Route (through Bale Highlands):")
    print(f"   {' → '.join(central_route)}")
    print(f"   Total cost: {central_cost}")

    if result['success']:
        print(f"\n🏆 A* found optimal path with cost: {result['cost']}")
        if result['cost'] <= min(southern_cost, central_cost):
            print("   ✓ A* found the optimal path")
        else:
            print("   ⚠ A* path might not be optimal (check heuristics)")


if __name__ == "__main__":
    main()
```

```python
class MiniMaxEthiopia:
    def __init__(self):
        # Tree structure from the figure
        self.tree = {
            "Addis Ababa": ["Ambo", "Buta Jirra", "Adama"],

            "Ambo": ["Nekemte", "Gedo"],
            "Nekemte": ["Gimbi", "Limu"],
            "Gedo": ["Shambu", "Fincha"],

            "Buta Jirra": ["Worabe", "Wolkite"],
            "Worabe": ["Hossana", "Durame"],
            "Wolkite": ["BenchNaji", "Tepi"],

            "Adama": ["Mojo", "DireDawa"],
            "Mojo": ["Kaffa", "Dilla"],
            "DireDawa": ["Chiro", "Harar"],
        }

        # Terminal utilities (coffee quality)
        self.utility = {
            "Shambu": 4, "Fincha": 5,
            "Gimbi": 8, "Limu": 8,
            "Hossana": 6, "Durame": 5,
            "BenchNaji": 5, "Tepi": 6,
            "Kaffa": 7, "Dilla": 9,
            "Chiro": 6, "Harar": 10,
        }

    def is_terminal(self, node):
        return node in self.utility

    def minimax(self, node, maximizing):
        if self.is_terminal(node):
            return self.utility[node], [node]

        if maximizing:  # Agent (MAX)
            max_eval = float('-inf')
            best_path = []
            for child in self.tree[node]:
                eval, path = self.minimax(child, False)
                if eval > max_eval:
                    max_eval = eval
                    best_path = [node] + path
            return max_eval, best_path

        else:  # Adversary (MIN)
            min_eval = float('inf')
            best_path = []
            for child in self.tree[node]:
                eval, path = self.minimax(child, True)
                if eval < min_eval:
                    min_eval = eval
                    best_path = [node] + path
            return min_eval, best_path


# Run MiniMax
game = MiniMaxEthiopia()
value, path = game.minimax("Addis Ababa", True)

print("Best achievable coffee quality:", value)
print("Optimal path:", " -> ".join(path))
```

```python
from graph_data import FIGURE_5_GRAPH
import rospy
from geometry_msgs.msg import Twist, Point
from nav_msgs.msg import Odometry
from sensor_msgs.msg import LaserScan
import math


class TravelingEthiopiaRobot:
    """Robot controller using BFS for pathfinding."""

    def __init__(self):
        """Initialize ROS node and setup."""
        rospy.init_node('traveling_ethiopia_robot', anonymous=True)

        # Publishers and Subscribers
        self.cmd_vel_pub = rospy.Publisher('/cmd_vel', Twist, queue_size=10)
        self.odom_sub = rospy.Subscriber('/odom', Odometry, self.odom_callback)
        self.laser_sub = rospy.Subscriber('/scan', LaserScan, self.laser_callback)

        # Robot state
        self.current_position = None
        self.current_orientation = None
        self.laser_data = None

        # Graph for pathfinding
        self.graph = FIGURE_5_GRAPH

        # City coordinates (Cartesian system for Gazebo world)
        # These would be defined in the .world file
        self.city_coordinates = self._initialize_city_coordinates()

        rospy.loginfo("Traveling Ethiopia Robot initialized")

    def _initialize_city_coordinates(self):
        """Set up city coordinates (should match the .world file)."""
        # Example coordinates - should match .world file
        coordinates = {
            'Addis Ababa': (0.0, 0.0, 0.0),
            'Adama': (2.0, 0.0, 0.0),
            'Ambo': (-2.0, 0.0, 0.0),
            'Wolkite': (0.0, 2.0, 0.0),
            'Jimma': (2.0, 2.0, 0.0),
            'Worabe': (0.0, 4.0, 0.0),
            'Hossana': (2.0, 4.0, 0.0),
            'Shashemene': (4.0, 4.0, 0.0),
            'Hawassa': (6.0, 4.0, 0.0),
            'Dilla': (8.0, 4.0, 0.0),
            # Add more cities as needed
        }
        return coordinates

    def odom_callback(self, msg):
        """Callback for odometry data."""
        self.current_position = msg.pose.pose.position
        orientation = msg.pose.pose.orientation
        # Convert quaternion to euler angles if needed
        self.current_orientation = orientation

    def laser_callback(self, msg):
        """Callback for laser scan data."""
        self.laser_data = msg

    def bfs_search(self, initial_state, goal_state):
        """BFS to find path from initial to goal city."""
        from collections import deque

        queue = deque([(initial_state, [initial_state])])
        visited = set()

        while queue:
            current, path = queue.popleft()

            if current == goal_state:
                return path

            if current not in visited:
                visited.add(current)

                for neighbor in self.graph.get(current, []):
                    if neighbor not in visited:
                        queue.append((neighbor, path + [neighbor]))

        return None

    def navigate_to_city(self, target_city):
        """Navigate to target city using pathfinding."""
        """
        if target_city not in self.city_coordinates:
            rospy.logwarn(f"City {target_city} not found in coordinates")
            return

        target_pos = self.city_coordinates[target_city]
        self.move_to_position(target_pos)

    def move_to_position(self, target_pos):
        """Move robot to target position."""
        rate = rospy.Rate(10)  # 10 Hz
        tolerance = 0.1

        while not rospy.is_shutdown():
            if self.current_position is None:
                rate.sleep()
                continue

            # Calculate distance and angle to target
            dx = target_pos[0] - self.current_position.x
            dy = target_pos[1] - self.current_position.y
            distance = math.sqrt(dx**2 + dy**2)

            if distance < tolerance:
                # Stop
                twist = Twist()
                self.cmd_vel_pub.publish(twist)
                rospy.loginfo("Reached target position")
                break

            # Simple proportional control
            angle = math.atan2(dy, dx)
            linear_speed = min(0.5, distance * 0.5)
            angular_speed = angle * 0.5

            twist = Twist()
            twist.linear.x = linear_speed
            twist.angular.z = angular_speed
            self.cmd_vel_pub.publish(twist)

            rate.sleep()

    def execute_path(self, initial_state, goal_state):
        """Execute complete path from initial to goal state."""
        rospy.loginfo(f"Finding path from {initial_state} to {goal_state}")
        path = self.bfs_search(initial_state, goal_state)

        if path is None:
            rospy.logerr("No path found")
            return

        rospy.loginfo(f"Path found: {' -> '.join(path)}")

        # Navigate through each city in path
        for city in path:
            rospy.loginfo(f"Navigating to {city}")
            self.navigate_to_city(city)
            rospy.sleep(1)  # Brief pause between cities

        rospy.loginfo("Path execution complete")


def main():
    """Main ROS node."""
    try:
        robot = TravelingEthiopiaRobot()

        # Example: Navigate from Addis Ababa to Jimma
        initial = 'Addis Ababa'
        goal = 'Jimma'

        robot.execute_path(initial, goal)

        rospy.spin()
    except rospy.ROSInterruptException:
        pass


if __name__ == "__main__":
    main()
```
